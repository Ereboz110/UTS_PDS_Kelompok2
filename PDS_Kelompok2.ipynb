```python
import numpy as np
import matplotlib.pyplot as plt

# Menentukan rentang nilai x (Server Standar)
x = np.linspace(0, 15, 400)

# Persamaan garis kendala:
# 1. 20x + 10y = 100  => y = (100 - 20x) / 10
y1 = (100 - 20*x) / 10

# 2. 10x + 30y = 90  => y = (90 - 10x) / 30
y2 = (90 - 10*x) / 30

#Titik garis dicari dengan substitusi 0 ke x dan y untuk mulai plotting garis pada bidang kartesius

# Membuat plot
plt.figure(figsize=(10, 8))
plt.plot(x, y1, label=r'$20x + 10y \geq 100$ (Pemrosesan Data)', color='blue', linewidth=2)
plt.plot(x, y2, label=r'$10x + 30y \geq 90$ (Akses Basis Data)', color='orange', linewidth=2)

# Titik potong (Intersection):
# Dari sistem persamaan:
# 20x + 10y = 100  (x3)  => 60x + 30y = 300
# 10x + 30y = 90   (x1)  => 10x + 30y = 90
# ------------------------------------------   -
#                           50x = 210  => x = 4.2
# Subtitusi x = 4.2 ke persamaan 1:
# 20(4.2) + 10y = 100 => 84 + 10y = 100 => 10y = 16 => y = 1.6
x_opt = 4.2
y_opt = 1.6

# Plot titik potong
plt.scatter(x_opt, y_opt, color='red', zorder=5, s=100, label=f'Titik Potong Optimal ({x_opt}, {y_opt})')
plt.annotate(f'({x_opt}, {y_opt})', (x_opt, y_opt), textcoords="offset points", xytext=(10,10), ha='center', fontweight='bold', color='red')

# Arsir Daerah Penyelesaian Layak
# Karena permasalahan mencari nilai minimum, maka arsir daerah diluar garis
y3 = np.maximum(np.maximum(y1, y2), 0)
plt.fill_between(x, y3, 15, where=(y3<=15), color='gray', alpha=0.3, label='Daerah Layak (Feasible Region)')

# Pengaturan grafik
plt.xlim(0, 10)
plt.ylim(0, 12)
plt.xlabel('Jam Server Standar (x)', fontsize=12)
plt.ylabel('Jam Server Performa Tinggi (y)', fontsize=12)
plt.title('Grafik Batasan Penyewaan Cloud Server', fontsize=14, fontweight='bold')
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(loc='upper right', fontsize=10)
plt.show()

# Menghitung biaya optimal di titik potong
biaya_optimal = 8000 * x_opt + 12000 * y_opt
print(f"--- Hasil Analisis ---")
print(f"Server Standar (x)          : {x_opt} jam")
print(f"Server Performa Tinggi (y)  : {y_opt} jam")
print(f"Biaya Sewa Minimum          : Rp. {biaya_optimal:,.0f}")
```
